## Final EDA Findings & Business Insights

### Key Findings

1. **Patient Demographics**
   - The dataset contains 1,000 patient appointment records.
   - The average patient age is approximately 53.75 years.
   - Female patients account for 52.7% of the records, while male patients account for 47.3%.
   - The 61–75 age group has the highest number of appointments.

2. **Department Performance**
   - Neurology has the highest number of appointments, indicating relatively high patient demand.
   - General department has the highest average billing amount at approximately 290.19.
   - Cardiology has the lowest average billing among the departments at approximately 259.55.

3. **Doctor Analysis**
   - Jennifer Adams has the highest number of appointments among the doctors.
   - Doctor-wise billing analysis helps identify differences in average revenue generated per doctor.

4. **Billing Analysis**
   - The average billing amount is approximately 275.80.
   - Billing amounts range from 51.38 to 499.75.
   - The billing distribution provides an overview of the revenue pattern across appointments.

5. **Follow-up Analysis**
   - 51.4% of appointments require follow-up, while 48.6% do not.
   - General has the highest follow-up percentage at approximately 54.11%.
   - Orthopedics has slightly more appointments without follow-up than with follow-up.

6. **Appointment Timing**
   - The average time between booking and appointment is approximately 274.56 days.
   - The minimum waiting time is 2 days, while the maximum is 728 days.
   - This variation suggests that some appointments are scheduled significantly farther in advance.

7. **Monthly Trends**
   - October 2025 recorded the highest number of appointments based on the appointment month analysis.
   - Monthly appointment trends can help healthcare organizations understand changes in patient demand.

### Business Insights

- Healthcare management can monitor departments with higher appointment demand to improve resource allocation.
- Billing patterns can help identify departments and doctors generating higher average revenue.
- Follow-up requirements can help hospitals plan future appointments and patient-care resources.
- Long appointment waiting periods may indicate opportunities to improve scheduling and capacity planning.
- Demographic and appointment trends can support better staffing and operational planning.
